```python
import numpy as np
import cv2
import tifffile
import matplotlib.pyplot as plt
import pathlib

def normalize_to_uint8(img):
    return cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

def remove_outliers(img, k=15.0, use_median=False):
    """
    Clip values that are more than k std-dev (or MAD units) above center.
    Args:
        img: 16-bit numpy array
        k: threshold (e.g. 3σ)
        use_median: if True use median+MAD, else mean+std
    Returns:
        clipped float32 image in [0,1]
    """
    x = img.astype(np.float32)

    if use_median:
        med = np.median(x)
        mad = np.median(np.abs(x - med)) + 1e-6
        sigma = 1.4826 * mad  # robust std estimate
        thresh = med + k * sigma
    else:
        mean = np.mean(x)
        std = np.std(x)
        thresh = mean + k * std

    # clip outliers
    x_clipped = np.minimum(x, thresh)

    # normalize after clipping (to 0..1)
    x_norm = (x_clipped - x_clipped.min()) / (x_clipped.max() - x_clipped.min() + 1e-6)
    return x_norm


# ----- AN DATA 2 -----

# 1
cyt_path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/data_vadym_singletif_STACKED/MAX_EXP_w488_s0026.tif"
nuc_path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/data_vadym_singletif_STACKED/MAX_EXP_wDAPI_s0026.tif"
cyt_img = tifffile.imread(cyt_path)
nuc_img = tifffile.imread(nuc_path)

smallest = np.amin(cyt_img)
biggest = np.amax(cyt_img)

# Before Removing Outliers
print("An Data 2 Original | Smallest 1:", smallest, "Biggest 1:", biggest)


# ----- OUTLIERS -----

st_dev = 15.0
cyt_removed = remove_outliers(cyt_img, st_dev)

smallest = np.amin(cyt_removed)
biggest = np.amax(cyt_removed)
print("An Data 2 Outliers Removed | Smallest 1:", smallest, "Biggest 1:", biggest)

# ----- NORMALIZED -----

cyt_norm = normalize_to_uint8(cyt_removed)

smallest = np.amin(cyt_norm)
biggest = np.amax(cyt_norm)
print("An Data 2 Outliers Removed + Normalized | Smallest 1:", smallest, "Biggest 1:", biggest)

# ----- VISUALIZATION -----
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 10), sharex=True, sharey=True)

ax[0].imshow(cyt_img, cmap='gray')
ax[0].set_title("Original")
ax[0].axis('off')

ax[1].imshow(cyt_removed, cmap='gray')
ax[1].set_title(f"Outliers Removed (k={st_dev})")
ax[1].axis('off')

ax[2].imshow(cyt_norm, cmap='gray')
ax[2].set_title(f"Normalized + Outliers Removed (k={st_dev})")
ax[2].axis('off')
```

```python
def remove_outliers(img, k=18.0, use_median=False, verbose=True, max_clip_frac=1e-4):
    """
    Clip values that are more than k std-dev (or MAD units) above center.
    Args:
        img: 16-bit numpy array
        k: threshold (e.g. 3σ)
        use_median: if True use median+MAD, else mean+std
    Returns:
        clipped float32 image in [0,1]
    """

    print("OUTLIERS REMOVING...")
    x = img.astype(np.float32)

    if use_median:
        med = np.median(x)
        mad = np.median(np.abs(x - med)) + 1e-6
        sigma = 1.4826 * mad  # robust std estimate
        thresh = med + k * sigma
    else:
        mean = np.mean(x)
        std = np.std(x)
        thresh = mean + k * std
    
    # ---- compute how many pixels would be clipped ----
    mask = x > thresh
    clip_frac = mask.mean()  # between 0 and 1

    if verbose:
        print(
            f"Threshold={thresh:.3f}, "
            f"clip_frac={clip_frac*100:.5f}% (max allowed={max_clip_frac*100:.5f}%)"
        )

    # ---- decide whether to actually clip ----
    if (clip_frac == 0) or (clip_frac > max_clip_frac):
        # either nothing to clip, or "too many" pixels above thresh
        # -> treat as no extreme outliers and skip clipping
        if verbose:
            if clip_frac == 0:
                print("No pixels above threshold — skipping outlier clipping.")
            else:
                print("Too many pixels above threshold — likely real signal, skipping clipping.")
        x_clipped = x
    else:
        # safe to treat as extreme outliers
        x_clipped = np.minimum(x, thresh)

    # ---- normalize after clipping (to 0..1) ----
    x_min = x_clipped.min()
    x_max = x_clipped.max()
    x_norm = (x_clipped - x_min) / (x_max - x_min + 1e-6)
    return x_norm


def clahe(img, clip_limit=4.0, tile_size=(8,8)):
    c = cv2.createCLAHE(clipLimit=clip_limit, tileGridSize=tile_size)
    return c.apply(img)


image_path = pathlib.Path("C:/Users/msgal/Downloads/Ding_Lab/image_testing/kayleigh_composite/488_pb5sp4_s1.composite.tif")
with tifffile.TiffFile(image_path) as tif:
    kayleigh_647 = tif.pages[0].asarray()      
    kayleigh_488 = tif.pages[1].asarray()   
    kayleigh_nuc = tif.pages[2].asarray() 

st_dev = 20.0

kayleigh_647_removed = remove_outliers(kayleigh_647, st_dev, use_median=False)
kayleigh_488_removed = remove_outliers(kayleigh_488, st_dev, use_median=False)

kayleigh_647_norm = normalize_to_uint8(kayleigh_647_removed)
kayleigh_488_norm = normalize_to_uint8(kayleigh_488_removed)

cyt_clahe_rem647 = clahe(kayleigh_647_norm, clip_limit=2.0)
cyt_clahe_raw647 = clahe(kayleigh_647, clip_limit=2.0)
cyt_clahe_488 = clahe(kayleigh_488_norm)

fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 10), sharex=True, sharey=True)

ax[0].imshow(cyt_clahe_488, cmap='gray')
ax[0].set_title("Channel 488 - CLAHE + Outliers Removed")
ax[0].axis('off')

ax[1].imshow(cyt_clahe_rem647, cmap='gray')
ax[1].set_title(f"Channel 647 - CLAHE + Outliers Removed (k={st_dev})")
ax[1].axis('off')

ax[2].imshow(cyt_clahe_raw647, cmap='gray')
ax[2].set_title(f"Channel 647 - CLAHE Only")
ax[2].axis('off')
```
